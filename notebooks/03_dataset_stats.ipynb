# 03 Dataset Stats

Use this notebook after labeling to inspect class balance, missing values, and split distribution.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd().resolve()
if ROOT.name == 'notebooks':
    ROOT = ROOT.parent

LABELS_PATH = ROOT / 'data' / 'labels' / 'critique_labels_v1.csv'
labels = pd.read_csv(LABELS_PATH).fillna('')
print(labels.shape)
labels.head()
```

```python
labels.info()
```

## Missing Values

```python
label_cols = ['composition', 'color_harmony', 'contrast', 'lighting', 'perspective']
missing_counts = labels[label_cols].eq('').sum().sort_values(ascending=False)
missing_counts
```

## Label Balance

```python
for col in label_cols:
    counts = labels[col].replace('', '<missing>').value_counts()
    ax = counts.plot(kind='bar', title=col, figsize=(6, 3))
    ax.set_xlabel('')
    ax.set_ylabel('rows')
    plt.show()
    display(counts.to_frame('count'))
```

## Split Distribution

```python
if 'split' in labels.columns:
    display(labels['split'].replace('', '<missing>').value_counts().to_frame('count'))
    pd.crosstab(labels['split'].replace('', '<missing>'), labels['composition'].replace('', '<missing>'))
```

## Metadata Coverage

```python
metadata_cols = ['source', 'artist', 'style', 'genre', 'emotion', 'description']
available = [c for c in metadata_cols if c in labels.columns]
coverage = {c: int(labels[c].replace('', pd.NA).notna().sum()) for c in available}
pd.Series(coverage).sort_values(ascending=False).to_frame('non_empty_rows')
```
